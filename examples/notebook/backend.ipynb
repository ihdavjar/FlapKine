```python
import numpy as np
import pandas as pd
from stl import mesh
from src.core.core import Scene, Sprite, Object3D
from src.core.transforms.flexibility import ConstantF, Flexibility_type1
from src.core.transforms.rotation import ConstantR, Rotation_EulerAngles
```

```python
stl_path = 'examples/Assets/STL/ellipse.stl'
your_mesh = mesh.Mesh.from_file(stl_path)

flexibility = Flexibility_type1(x=True, y=True, z=True, major_axis=5, minor_axis=3)
rotation = Rotation_EulerAngles(type='ZYX')

ob = Object3D(name='RightWing', stl_mesh=your_mesh, translation_transform=flexibility, rotation_transform=rotation)

angle_alpha = pd.read_csv('examples/Assets/Angles/data_alpha.csv')
angle_beta = pd.read_csv('examples/Assets/Angles/data_beta.csv')
angle_gamma = pd.read_csv('examples/Assets/Angles/data_gamma.csv')

angle_alpha = np.array(angle_alpha)
angle_beta = np.array(angle_beta)
angle_gamma = np.array(angle_gamma)

angles = np.concatenate((angle_alpha, angle_beta, angle_gamma), axis=1)

sprite = Sprite(ob, angles)

scene = Scene([sprite])

scene.save_stl(t=90, path ='ellipse.stl')

```

```python
# A function to plot stl image using plotly
def plot_stl(stl_path):
    import plotly.graph_objects as go
    import numpy as np
    from stl import mesh

    # Using an existing stl file:
    mesh = mesh.Mesh.from_file(stl_path)
    x, y, z = mesh.x, mesh.y, mesh.z

    fig = go.Figure(data=[go.Mesh3d(x=x, y=y, z=z, color='lightpink', opacity=1)])
    fig.show()
```
